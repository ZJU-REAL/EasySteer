# Controlling Thinking Speed — steering replication

Replicates the thinking-speed steering from **"Controlling Thinking Speed in Reasoning Models"** ([arXiv:2507.03704](http://arxiv.org/abs/2507.03704)) on DeepSeek-R1-Distill-Qwen-1.5B.

The slow→fast direction (`MATH500.gguf`, built in `data_construction.ipynb`) is added at layers 19–27 — the paper's control range for this 28-layer architecture — during the generation of every token. A positive scale enhances fast thinking, cutting the reasoning tokens substantially while preserving the boxed answer.


```python
import os

from vllm import LLM, SamplingParams
from vllm.steer_vectors import ApplySpec, SteeringSpec, VectorSpec

os.environ.setdefault("CUDA_VISIBLE_DEVICES", "0")

MODEL = "/home/shenyl/hf/model/deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B/"  # deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B

llm = LLM(
    model=MODEL,
    enable_steer_vector=True,
    enforce_eager=True,
)
tok = llm.get_tokenizer()
```

```python
INSTRUCTION = "Please reason step by step, and put your final answer within \\boxed{}."
problem = "Find the constant term in the expansion of $$\\left(10x^3-\\frac{1}{2x^2}\\right)^{5}$$"

text = tok.apply_chat_template(
    [{"role": "user", "content": f"{INSTRUCTION}\n{problem}"}],
    tokenize=False,
    add_generation_prompt=True,
)
ids = tok(text, add_special_tokens=False).input_ids
params = SamplingParams(temperature=0, max_tokens=8192, skip_special_tokens=False)

baseline = llm.generate({"prompt_token_ids": ids}, params)
print(baseline[0].outputs[0].text)
print("Baseline tokens:", len(baseline[0].outputs[0].token_ids))
```

```python
# Fast thinking: positive scale on the slow->fast direction at the
# paper's control layers (last 10 except the final layer).
steering_fast = SteeringSpec(vectors=[
    VectorSpec(
        source="MATH500.gguf",
        scale=4.0,
        layers=list(range(19, 28)),
        apply=ApplySpec(phases=["prompt", "generation"]),
    ),
])
fast = llm.generate({"prompt_token_ids": ids}, params, steering=steering_fast)
print(fast[0].outputs[0].text)
print("Fast tokens:", len(fast[0].outputs[0].token_ids))
```
