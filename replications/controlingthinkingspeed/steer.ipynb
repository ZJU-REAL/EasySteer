# Controlling Thinking Speed — steering replication

Replicates the thinking-speed steering from **"Controlling Thinking Speed in Reasoning Models"** ([arXiv:2507.03704](http://arxiv.org/abs/2507.03704)) on DeepSeek-R1-Distill-Qwen-1.5B.

The slow-fast thinking direction is extracted from reasoning traces primed with "To" (fast) vs "Alright" (slow) continuations (see `data_construction.ipynb`, which produces `MATH500.gguf`). Applied at the upper layers (19-27) with a positive scale it pushes the model toward faster thinking, cutting reasoning tokens; this notebook contrasts that against an unsteered baseline.

Uses the EasySteer v2 steering API (`SteeringSpec` / `VectorSpec` / `ApplySpec`).

```python
import os

from transformers import AutoTokenizer

from vllm import LLM, SamplingParams
from vllm.steer_vectors import ApplySpec, SteeringSpec, VectorSpec

os.environ.setdefault("CUDA_VISIBLE_DEVICES", "0")

MODEL = "/home/shenyl/hf/model/deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B/"  # deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B

llm = LLM(
    model=MODEL,
    enable_steer_vector=True,
    enforce_eager=True,
)
tokenizer = AutoTokenizer.from_pretrained(MODEL)
```

```python
# Baseline: no steering (the model reasons slowly / at length).
prompt_template = "<|User|>Return your final response within \\boxed{}.\n%s<|Assistant|><think>\n"
prompt = "Find the constant term in the expansion of $$\\left(10x^3-\\frac{1}{2x^2}\\right)^{5}$$"
text = prompt_template % prompt
params = SamplingParams(temperature=0, max_tokens=4096, skip_special_tokens=False)

output = llm.generate(text, params)
print(output[0].outputs[0].text)
print("Baseline tokens: ", len(tokenizer.tokenize(output[0].outputs[0].text, add_special_tokens=True)))
```

```python
# Steer toward fast thinking: apply the direction at the upper layers
# (19-27) across prompt and generation. Fewer reasoning tokens result.
steering_fast = SteeringSpec(
    vectors=[
        VectorSpec(
            source="MATH500.gguf",
            scale=4.0,
            layers=list(range(19, 28)),
            apply=ApplySpec(phases=["prompt", "generation"]),
        )
    ],
)
output = llm.generate(text, params, steering=steering_fast)
print(output[0].outputs[0].text)
print("Fast tokens: ", len(tokenizer.tokenize(output[0].outputs[0].text, add_special_tokens=True)))
```
