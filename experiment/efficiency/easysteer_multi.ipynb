```python
# Import necessary libraries
import os
from vllm import LLM, SamplingParams
from vllm.steer_vectors import ApplySpec, SteeringSpec, VectorSpec
from transformers import AutoTokenizer

# Set environment variables
os.environ.setdefault("CUDA_VISIBLE_DEVICES", "0")

# Initialize LLM with steering vector capability
llm = LLM(
    model="/home/shenyl/hf/model/deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B/",
    enable_steer_vector=True,
    enforce_eager=True,
    tensor_parallel_size=1
)
```

```python
import json
file_path = "/home/xhl/eval/my_eval/data/math/train.jsonl"

problems = []
answers = []

with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        item = json.loads(line)
        problems.append(item["problem"])
        answers.append(item["answer"])
examples = ["Please reason step by step, and put your final answer within \\boxed{}.\nUser: " + prompt + "\nAssistant: <think>" for prompt in problems]
```

```python
steering = SteeringSpec(vectors=[
    VectorSpec(
        source="execution_avg_vector.gguf",
        scale=0.0,
        algorithm="direct",
        # layers=[20],
        apply=ApplySpec(phases=["prompt", "generation"]),
    ),
])
# Multi-vector spec for the throughput benchmark: three copies of the same
# vector at scale 0.0, applied on all 28 layers, so the multi-vector steering
# path is exercised without changing the generated text.
steering = SteeringSpec(
    vectors=[
        VectorSpec(
            source="execution_avg_vector.gguf",
            scale=0.0,
            layers=list(range(28)),
            algorithm="direct",
            normalize=False,
            apply=ApplySpec(phases=["prompt", "generation"]),
        ),
        VectorSpec(
            source="execution_avg_vector.gguf",
            scale=0.0,
            layers=list(range(28)),
            algorithm="direct",
            normalize=False,
            apply=ApplySpec(phases=["prompt", "generation"]),
        ),
        VectorSpec(
            source="execution_avg_vector.gguf",
            scale=0.0,
            layers=list(range(28)),
            algorithm="direct",
            normalize=False,
            apply=ApplySpec(phases=["prompt", "generation"]),
        ),
    ],
    debug=False,              # Don't output debug info
    conflict="sequential",    # Apply vectors in sequence
)

import time
total_output_tokens=0
start_time = time.time()
for i in range(10):
    example_answers = llm.generate(
        examples[i],
        SamplingParams(
            temperature=0,
            max_tokens=2048,
            skip_special_tokens=False,
        ),
        steering=steering
    )
    total_output_tokens += sum((len(r.outputs[0].token_ids) if r.outputs else 0) for r in example_answers)
end_time = time.time()
toks_per_sec = total_output_tokens / (end_time - start_time) if end_time - start_time > 0 else 0
print(f"{(end_time - start_time)/10:.8f}s/req")
print(f"output toks/s: {toks_per_sec:.2f}")

# batch
# example_answers = llm.generate(
#     examples[:512],
#     SamplingParams(
#         temperature=0,
#         max_tokens=2048,
#         skip_special_tokens=False,
#     ),
#     steering=steering
# )
```
